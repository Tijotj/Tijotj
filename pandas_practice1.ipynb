```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
display(data.head(10))
```

```python
data = data[data['Deposit Amt.'].notna()]
display(data.head())
```

```python
data = data.iloc[1:].copy()
display(data.head())
```

```python
sreehari_store_data = data[data['Narration'].str.contains('Sreehari Store', na=False, case=False)]
#display(sreehari_store_data.head())
sreehari_store_data
```

```python
total_sreehari_withdrawal = sreehari_store_data['Withdrawal Amt.'].sum()
print(f"Total Withdrawal Amount for Sreehari Store: {total_sreehari_withdrawal}")
```

```python
fuel_data = data[data['Narration'].str.contains('Fuel', na=False, case=False)]
total_fuel_withdrawal = fuel_data['Withdrawal Amt.'].sum()
print(f"Total Withdrawal Amount for Fuel transactions: {total_fuel_withdrawal}")
```

```python
swiggy_data = data[data['Narration'].str.contains('Swiggy', na=False, case=False)]
total_swiggy_deposit = swiggy_data['Deposit Amt.'].sum()
print(f"Total Deposit Amount for Swiggy transactions: {total_swiggy_deposit}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a DataFrame for comparison
comparison_data = pd.DataFrame({
    'Category': ['Sreehari Store (Withdrawal)', 'Fuel (Withdrawal)', 'Swiggy (Deposit)'],
    'Total Amount': [total_sreehari_withdrawal, total_fuel_withdrawal, total_swiggy_deposit]
})

# Plotting the bar chart
fig = plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Total Amount', data=comparison_data, palette='viridis')
plt.title('Comparison of Total Amounts: Sreehari Store, Fuel, and Swiggy')
plt.ylabel('Total Amount (₹)')
plt.xlabel('Category')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
